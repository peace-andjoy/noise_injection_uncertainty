```python
import pandas as pd
```

```python
result_df = []
# 打开文件
with open('evaluation.log', 'r') as f:
    # 逐行读取文件内容
    for line in f:
        # 去除行尾的换行符并打印每一行
        infos = line.replace(' ', '').replace('(', '').replace(')', '').strip().split('{')
        li = infos[0].split("'")
        # time = li[0]
        dataset_name = li[1]
        model_name = li[3]
        batch_size, weight_decay, learning_rate = [l for l in li[4].split(',') if l]
        tmp = infos[1].split('}')
        epoch2loss = tmp[0]
        best_epoch, best_loss, fix_noiselevel, init_noiselevel, dropout_prob = [s for s in tmp[1].split(',') if s]
        result_df.append((dataset_name, model_name, batch_size, weight_decay, learning_rate, epoch2loss, best_epoch, best_loss, fix_noiselevel, init_noiselevel, dropout_prob))
    
```

```python
result_df = pd.DataFrame(result_df, columns=['dataset_name', 'model_name', 'batch_size', 'weight_decay', 'learning_rate', 'epoch2loss', 'best_epoch', 'best_loss', 'fix_noiselevel', 'init_noiselevel', 'dropout_prob'])
```

```python
result_df['batch_size'] = result_df['batch_size'].astype(int)
result_df['weight_decay'] = result_df['weight_decay'].astype(float)
result_df['learning_rate'] = result_df['learning_rate'].astype(float)
result_df['best_epoch'] = result_df['best_epoch'].astype(int)
result_df['best_loss'] = result_df['best_loss'].astype(float)
result_df['fix_noiselevel'] = result_df['fix_noiselevel'].astype(float)
result_df['init_noiselevel'] = result_df['init_noiselevel'].astype(float)
result_df['dropout_prob'] = result_df['dropout_prob'].astype(float)
```

```python
FCN_result_df = result_df[result_df['model_name']=='FCN']
noisyFCN_result_df = result_df[result_df['model_name']=='noisyFCN']
fixnoiseFCN_result_df = result_df[result_df['model_name']=='fixnoiseFCN']
dropoutFCN_result_df = result_df[result_df['model_name']=='FCN']
```

```python
noisyFCN_result_df = noisyFCN_result_df.loc[noisyFCN_result_df.groupby(['dataset_name', 'batch_size', 'weight_decay', 'learning_rate'])['best_loss'].idxmin().reset_index(drop=True)]
fixnoiseFCN_result_df = fixnoiseFCN_result_df.loc[fixnoiseFCN_result_df.groupby(['dataset_name', 'batch_size', 'weight_decay', 'learning_rate'])['best_loss'].idxmin().reset_index(drop=True)]
dropoutFCN_result_df = dropoutFCN_result_df.loc[dropoutFCN_result_df.groupby(['dataset_name', 'batch_size', 'weight_decay', 'learning_rate'])['best_loss'].idxmin().reset_index(drop=True)]
```

```python
df1 = pd.merge(FCN_result_df, noisyFCN_result_df, on=['dataset_name', 'batch_size', 'weight_decay', 'learning_rate'], how='inner', suffixes=('_FCN', '_noisyFCN'))
df2 = pd.merge(df1, fixnoiseFCN_result_df, on=['dataset_name', 'batch_size', 'weight_decay', 'learning_rate'], how='inner')
df3 = pd.merge(df2, dropoutFCN_result_df, on=['dataset_name', 'batch_size', 'weight_decay', 'learning_rate'], how='inner', suffixes=('_fixnoiseFCN', '_dropoutFCN'))
df3
```
